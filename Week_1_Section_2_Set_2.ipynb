```python
import pandas as pd
import numpy as np

data = {
"Name": ["Alice", "Bob", "Charlie", "David", "Eva"],
"Age": [25, np.nan, 35, 40, None],
"City": ["NYC", "LA", None, "Miami", "Boston"],
}

df = pd.DataFrame(data)
```

Exercise 1 - View the first 5 rows of the DataFram

```python
df.head()
```

Excercise 2 - Show where the values are missing.

```python
df.isnull()
```

Excercise 3 - Count missing values by column

```python
df.isnull().sum()
```

Excercise 4 - Drop rows with missing data.

```python
df = df.dropna()
print(df)
```

Excercise 5 - Drop columns with missing data.

```python
df = df.dropna(axis=1)
print(df)
```

Excersise 6 - Fill missing Age values with the mean.

```python
df = df.fillna(df['Age'].mean())
print(df)
```

Excersise 7 - Fill missing City values with `Unknown`

```python
df['City'] = df['City'].fillna('Unknown')
print(df)
```
