```python
import pandas as pd

data = {
"Name": ["Alice", "Bob", "Charlie", "Alice", "Bob", "Diana"],
"Age": [25, 30, 35, 25, 30, 40],
"City": ["NYC", "LA", "SF", "NYC", "LA", "NYC"],
}

df = pd.DataFrame(data)
```

Exercise 1 - Find which rows are duplicates across all columns.

```python
df.duplicated()
```

Exercise 2 - Drop duplicate rows, keeping the first occurrence.

```python
df_cleaned = df.drop_duplicates()
df_cleaned
```

Excersise 3 - Drop duplicate rows, keeping only the last occurrence.

```python
df_cleaned = df.drop_duplicates(keep='last')
df_cleaned
```

Excersise 4 - Drop all duplicates, removing the original and duplicate rows.

```python
df_cleaned = df.drop_duplicates(keep=False)
df_cleaned
```
