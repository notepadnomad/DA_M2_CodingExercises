```python
import pandas as pd

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)

print(df.head())
```

```python
# Preview the data
print(df.head())
```

```python
# Use .loc to get the name and age of the passenger with index 12
df.loc[12, ['Name', 'Age']]
```

```python
# Use .iloc to get the first 5 rows and first 3 columns
df.iloc[0:5, 0:3]
```

```python
# Use .loc to get all passengers in 1st class who are female
print(df.loc[(df['Pclass'] == 1) & (df['Sex'] == 'female')])
```

```python
# Use .iloc to get the age od the 10th passenger (by position)
print(df.iloc[9, 5])
```

```python
# Use .loc to get the 'Fare' and 'Survived' values for passengers with index 100 to 105
print(df.loc[100:105, ['Fare', 'Survived']])
```
