```python
import pandas as pd

data = {
"Store": ["A", "A", "A", "B", "B", "C", "C", "C", "C"],
"Product": ["Apples", "Oranges", "Bananas", "Apples", "Bananas", "Oranges", "Apples", "Bananas", "Bananas"],
"Sales": [100, 80, 90, 120, 110, 60, 95, 85, 75],
"Profit": [30, 25, 20, 40, 35, 15, 32, 22, 18],
}

df = pd.DataFrame(data)
print(df.head())
```

Excercise 1 - Group the data by `Store` and use `.agg()` to find the aveage `Sales` and average `Profit` for each store.

```python
df.groupby('Store').agg({
    'Sales': 'mean',
    'Profit':'mean'
})
```

Excercise 2 - Group by `Product` and get the maximum `Sales` and minimum `Profit`

```python
df.groupby('Product').agg({
    'Sales': 'max',
    'Profit': 'min'
})
```

Excercise 3 - Group by `Store` and return the average `Sales` and maximum `Profit`, but name the columns `Avg_Sales` and `Max_Profit`.

```python
df.groupby('Store').agg(
    Avg_Sales=('Sales', 'mean'),
    Max_Profit=('Profit', 'max')
)
```

BONUS: Group by both `Store` and `Product`, and find the total `Sales` using `.agg()`.

```python
df.groupby(['Store', 'Product']).agg({
    'Sales': 'sum'
})
```
