```python
import pandas as pd

df_customers = pd.DataFrame({
"CustomerName": ["Alice", "Bob", "Charlie"]
}, index=[101, 102, 103])

df_orders = pd.DataFrame({
"OrderTotal": [250, 300, 400]
}, index=[101, 102, 103])

df_cities = pd.DataFrame({
"Cities": ["NYC", "LA", "Chicago"]
}, index=[101, 102, 103])
```

Exercise 1 - Join `df_orders` to `df_customers` on the index.

```python
df_left = pd.DataFrame({'CustomerName': ["Alice", "Bob", "Charlie"]},
                       index=[101, 102, 103])

df_right = pd.DataFrame({'OrderTotal': [250, 300, 400]},
                        index=[101, 102, 103])

result = df_left.join(df_right)
print(result)
```
